```python

# Exercise 1: Engineering Test Data Organizer

LAST_NAME = "Nicolas"
SEED_NUM = 1
FAVORITE_ARTIST = "NIKI"

categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]

# Generate individualized measurement records
records = []

for i, source in enumerate(test_sources):
    artist_char = FAVORITE_ARTIST[i % len(FAVORITE_ARTIST)]
    value = (ord(artist_char) + SEED_NUM + i) % 100

    records.append({
        "source": source,
        "category": categories[i],
        "value": value
    })

# Add multiple measurements and one repeated source/category record
records.extend([
    {"source": test_sources[0], "category": "TEMPERATURE", "value": 25},
    {"source": test_sources[1], "category": "VOLTAGE", "value": 12},
    records[0].copy()
])

# 1. Display generated data
print("=== GENERATED TEST DATA ===")
for record in records:
    print(record)

# 2. Organize records by source and category
organized_data = {}
for record in records:
    source = record["source"]
    category = record["category"]
    organized_data.setdefault(source, {}).setdefault(category, []).append(record["value"])

print("\n=== ORGANIZED TEST DATA ===")
for source, measurements in organized_data.items():
    print(source, ":", measurements)

# 3. Identify repeated and distinct source/category pairs
pairs = [(r["source"], r["category"]) for r in records]
unique_pairs = set(pairs)
repeated_pairs = {pair for pair in unique_pairs if pairs.count(pair) > 1}

print("\n=== REPEATED / DISTINCT DATA ===")
print("Repeated Source/Category Pairs:", repeated_pairs)
print("Distinct Source/Category Pairs:", unique_pairs)

# 4. Numerical and categorical summaries
values = [record["value"] for record in records]
category_counts = {}

for record in records:
    category_counts[record["category"]] = (
        category_counts.get(record["category"], 0) + 1
    )

print("\n=== NUMERICAL AND CATEGORICAL SUMMARIES ===")
print("Measurement Count:", len(values))
print("Minimum:", min(values))
print("Maximum:", max(values))
print("Average:", round(sum(values) / len(values), 2))
print("Category Counts:", category_counts)

# 5. Sort records by source and category
processed_results = sorted(
    records, key=lambda r: (r["source"], r["category"], r["value"])
)

print("\n=== ORDERED RESULTS ===")
for record in processed_results:
    print(record)

# 6. Final consolidated summary
print("\n=== FINAL ORGANIZED AND SUMMARIZED TEST DATASET ===")
print("Test Sources:", test_sources)
print("Measurement Categories:", categories)
print("Total Records:", len(records))
print("Unique Source/Category Pairs:", len(unique_pairs))
print("Repeated Source/Category Pairs:", len(repeated_pairs))
print("Overall Average:", round(sum(values) / len(values), 2))
```

```python

# APPENDIX EXERCISE 2: STUDENT PERFORMANCE DATA ANALYZER

LAST_NAME = "Nicolas"
SEED_NUM = 1
FAVORITE_ARTIST = "NIKI"

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

student_labels = ["S1", "S2", "S3", "S4", "S5"]
students = [f"{LAST_NAME}_{label}" for label in student_labels]

# Generate scores using SEED_NUM and character positions of FAVORITE_ARTIST
records = []

for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        character_value = ord(FAVORITE_ARTIST[(i + j) % len(FAVORITE_ARTIST)])
        score = (character_value + SEED_NUM + i * 7 + j * 3) % 41 + 60

        records.append({
            "student": student,
            "subject": subject,
            "score": score
        })

# Add one repeated student-subject record
records.append(records[0].copy())

# Make one existing student-subject record incomplete
for record in records:
    if record["student"] == students[4] and record["subject"] == "ENGLISH":
        record["score"] = None
        break

print("=== GENERATED PERFORMANCE RECORDS ===")
for record in records:
    print(record)

# Organize records by student and subject
organized_data = {}

for record in records:
    student = record["student"]
    subject = record["subject"]
    score = record["score"]

    organized_data.setdefault(student, {}).setdefault(subject, []).append(score)

print("\n=== ORGANIZED STUDENT/SUBJECT DATA ===")
for student, subject_data in organized_data.items():
    print(student, ":", subject_data)

# Identify repeated and incomplete records
record_counts = {}
for record in records:
    key = (record["student"], record["subject"])
    record_counts[key] = record_counts.get(key, 0) + 1

repeated_records = {
    key: count for key, count in record_counts.items() if count > 1
}
incomplete_records = [
    record for record in records if record["score"] is None
]

print("\n=== REPEATED / INCOMPLETE RECORDS ===")
print("Repeated Student/Subject Records:", repeated_records)
print("Incomplete Records:", incomplete_records)

# Assign performance categories to valid scores
def performance_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 75:
        return "SATISFACTORY"
    else:
        return "NEEDS IMPROVEMENT"

valid_records = []
categorized_records = []

for record in records:
    category = performance_category(record["score"])
    categorized_record = {**record, "category": category}
    categorized_records.append(categorized_record)

    if record["score"] is not None:
        valid_records.append(record)

print("\n=== PERFORMANCE CATEGORIES ===")
for record in categorized_records:
    print(record)

# Summaries by student
print("\n=== STUDENT PERFORMANCE SUMMARIES ===")
for student in students:
    scores = [
        record["score"] for record in valid_records
        if record["student"] == student
    ]
    if scores:
        print(
            student,
            "| Valid Scores:", len(scores),
            "| Average:", round(sum(scores) / len(scores), 2),
            "| Highest:", max(scores),
            "| Lowest:", min(scores)
        )

# Summaries by subject
print("\n=== SUBJECT PERFORMANCE SUMMARIES ===")
for subject in subjects:
    scores = [
        record["score"] for record in valid_records
        if record["subject"] == subject
    ]
    if scores:
        print(
            subject,
            "| Valid Scores:", len(scores),
            "| Average:", round(sum(scores) / len(scores), 2),
            "| Highest:", max(scores),
            "| Lowest:", min(scores)
        )

# Order valid results from highest to lowest score
ordered_results = sorted(
    valid_records,
    key=lambda record: record["score"],
    reverse=True
)

print("\n=== ORDERED RESULTS (HIGHEST TO LOWEST) ===")
for record in ordered_results:
    print(record)

# Consolidated final report
all_scores = [record["score"] for record in valid_records]

print("\n=== FINAL STUDENT PERFORMANCE REPORT ===")
print("Total Records:", len(records))
print("Valid Scores:", len(valid_records))
print("Incomplete Records:", len(incomplete_records))
print("Repeated Student/Subject Pairs:", len(repeated_records))
print("Unique Students:", len(set(record["student"] for record in records)))
print("Subjects Analyzed:", len(set(record["subject"] for record in records)))
print("Overall Average:", round(sum(all_scores) / len(all_scores), 2))
print("Highest Score:", max(all_scores))
print("Lowest Score:", min(all_scores))
```

```python

# APPENDIX EXERCISE 3: TRANSACTION HISTORY ANALYZER

LAST_NAME = "Nicolas"
SEED_NUM = 1
FAVORITE_ARTIST = "NIKI"

products = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 250,
    "HEADPHONES": 500,
    "CALCULATOR": 350
}

customer_labels = ["C1", "C2", "C3", "C4", "C5"]
customers = [f"{LAST_NAME}_{label}" for label in customer_labels]
product_names = list(products.keys())

# Generate transaction records using SEED_NUM and FAVORITE_ARTIST
transactions = []

for i, customer in enumerate(customers):
    for j, product in enumerate(product_names):
        character_value = ord(FAVORITE_ARTIST[(i + j) % len(FAVORITE_ARTIST)])
        quantity = (character_value + SEED_NUM + i + j) % 5 + 1

        transactions.append({
            "customer": customer,
            "product": product,
            "quantity": quantity,
            "unit_price": products[product]
        })

# Add one repeated customer-product transaction
transactions.append(transactions[0].copy())

# Calculate each transaction's value
for transaction in transactions:
    transaction["transaction_value"] = (
        transaction["quantity"] * transaction["unit_price"]
    )

print("=== GENERATED TRANSACTION RECORDS ===")
for transaction in transactions:
    print(transaction)

# Organize transactions by customer and product
organized_transactions = {}

for transaction in transactions:
    customer = transaction["customer"]
    product = transaction["product"]

    organized_transactions.setdefault(customer, {}).setdefault(
        product, []
    ).append(transaction)

print("\n=== ORGANIZED TRANSACTION DATA ===")
for customer, product_data in organized_transactions.items():
    print(f"\n{customer}:")
    for product, records in product_data.items():
        print(product, ":", records)

# Identify repeated and distinct customer-product pairs
pair_counts = {}

for transaction in transactions:
    pair = (transaction["customer"], transaction["product"])
    pair_counts[pair] = pair_counts.get(pair, 0) + 1

repeated_transactions = {
    pair: count for pair, count in pair_counts.items() if count > 1
}

distinct_products = set(
    transaction["product"] for transaction in transactions
)
distinct_customers = set(
    transaction["customer"] for transaction in transactions
)

print("\n=== REPEATED / DISTINCT TRANSACTIONS ===")
print("Repeated Customer/Product Pairs:", repeated_transactions)
print("Distinct Products:", sorted(distinct_products))
print("Distinct Customers:", sorted(distinct_customers))

# Display transaction values
print("\n=== TRANSACTION VALUES ===")
for transaction in transactions:
    print(
        transaction["customer"],
        "|", transaction["product"],
        "| Quantity:", transaction["quantity"],
        "| Unit Price:", transaction["unit_price"],
        "| Transaction Value:", transaction["transaction_value"]
    )

# Summaries by customer
print("\n=== CUSTOMER SUMMARIES ===")
for customer in customers:
    customer_transactions = [
        transaction for transaction in transactions
        if transaction["customer"] == customer
    ]
    total = sum(
        transaction["transaction_value"]
        for transaction in customer_transactions
    )
    print(
        customer,
        "| Transactions:", len(customer_transactions),
        "| Total Spending:", total
    )

# Summaries by product
print("\n=== PRODUCT SUMMARIES ===")
for product in product_names:
    product_transactions = [
        transaction for transaction in transactions
        if transaction["product"] == product
    ]
    total_quantity = sum(
        transaction["quantity"] for transaction in product_transactions
    )
    total_value = sum(
        transaction["transaction_value"]
        for transaction in product_transactions
    )
    print(
        product,
        "| Total Quantity:", total_quantity,
        "| Total Sales:", total_value
    )

# Order transactions from highest to lowest value
ordered_transactions = sorted(
    transactions,
    key=lambda transaction: transaction["transaction_value"],
    reverse=True
)

print("\n=== ORDERED RESULTS (HIGHEST TO LOWEST VALUE) ===")
for transaction in ordered_transactions:
    print(transaction)

# Final consolidated transaction report
overall_value = sum(
    transaction["transaction_value"] for transaction in transactions
)

print("\n=== FINAL CONSOLIDATED TRANSACTION REPORT ===")
print("Total Transaction Records:", len(transactions))
print("Unique Customers:", len(distinct_customers))
print("Distinct Products:", len(distinct_products))
print("Repeated Customer/Product Pairs:", len(repeated_transactions))
print("Overall Transaction Value:", overall_value)
print("Total Units Sold:", sum(t["quantity"] for t in transactions))
```
